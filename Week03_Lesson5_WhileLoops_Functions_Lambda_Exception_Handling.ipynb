## 2. while Loops

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A **while loop** repeats a block of code for as long as a condition remains `True`.
Unlike a `for` loop (which runs once per list item), a `while` loop keeps going until *you* tell it to stop.

```python
while condition:
    # this block runs as long as condition is True
    # update something so condition eventually becomes False!
```

| Situation | Best choice |
|-----------|-------------|
| Process every item in a list | `for` loop |
| Know exactly how many steps | `for` loop |
| Keep going until something changes | `while` loop |
| Search for something and stop when found | `while` loop |
| Simulate a process with an unknown end point | `while` loop |

</div>

````python
# Simple while loop: count from 1 to 5
count = 1

while count <= 5:
    print(f'Step {count}')
    count = count + 1   # IMPORTANT: update count or the loop runs forever!

print('Loop finished.')
````

````python
# Simulate drilling: keep going until target depth is reached
current_depth = 1000.0   # metres
target_depth  = 1005.0   # metres
drill_rate    = 0.5      # metres per step

print('--- Drilling Report ---')
while current_depth < target_depth:
    print(f'  Drilling at {current_depth:.1f} m...')
    current_depth = current_depth + drill_rate

print(f'Target depth {target_depth} m reached. Drilling complete.')
````

````python
# Search GR log for first clean sand interval (GR < 50 API)
i = 0
sand_found = False

while i < len(gr_log):
    if gr_log[i] < 50:
        print(f'First clean sand at {depth_log[i]:.1f} m -- GR = {gr_log[i]} API')
        sand_found = True
        break       # exit the loop immediately
    i = i + 1

if not sand_found:
    print('No clean sand found in this log.')
````

> **Caution -- Infinite Loops**
>
> A while loop whose condition *never* becomes `False` will run forever and freeze your notebook.
> Always make sure something **inside the loop** updates the condition.
>
> ```python
> # BAD: depth never changes -- this runs forever
> depth = 1000.0
> while depth < 2000.0:
>     print("stuck!")
>
> # GOOD: depth updates each iteration
> depth = 1000.0
> while depth < 2000.0:
>     print(f"At {depth:.1f} m")
>     depth = depth + 10.0
> ```
>
> If you create an infinite loop by accident, click the **Stop** (square &#9632;) button in the toolbar.

### 3a. Functions with Multiple Parameters

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A function can accept **any number of inputs** separated by commas.
When you call the function, supply one value per parameter in the same order.

```python
def function_name(param1, param2, param3):
    # use all three inputs
    return result

function_name(val1, val2, val3)   # call with three arguments
```

</div>

````python
# Function with three parameters
def net_pay(thickness, porosity, sw):
    """
    Estimate net pay from petrophysical parameters.

    thickness : gross interval thickness in metres
    porosity  : porosity fraction (e.g. 0.28 = 28 %)
    sw        : water saturation fraction (e.g. 0.35 = 35 %)
    Returns   : net pay in metres
    """
    pore_volume_frac = porosity * (1 - sw)
    return thickness * pore_volume_frac

# Base case
print(f'Base case:              {net_pay(10.0, 0.28, 0.35):.2f} m net pay')

# Sensitivity: change one parameter at a time
print(f'Higher porosity (0.35): {net_pay(10.0, 0.35, 0.35):.2f} m net pay')
print(f'Lower Sw (0.20):        {net_pay(10.0, 0.28, 0.20):.2f} m net pay')
print(f'Thicker interval (15m): {net_pay(15.0, 0.28, 0.35):.2f} m net pay')
````

### 3b. Default Parameter Values

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

Give a parameter a **default value** so the caller does not have to supply it every time.

```python
def convert(value, unit="feet"):   # "feet" is the default
    ...

convert(2000)            # uses default -- feet
convert(2000, "km")      # overrides with km
convert(2000, unit="km") # keyword argument -- same result, more readable
```

Default parameters must come **after** any non-default parameters.

</div>

````python
def depth_convert(depth_m, unit='feet'):
    """
    Convert a depth from metres to another unit.
    unit : 'feet' (default), 'km', or 'm' (no conversion)
    """
    if unit == 'feet':
        return depth_m * 3.28084
    elif unit == 'km':
        return depth_m / 1000.0
    else:
        return depth_m

# Default (feet)
print(f'2000 m = {depth_convert(2000):.2f} ft   (default unit)')

# Override
print(f'2000 m = {depth_convert(2000, "km"):.3f} km')

# Keyword argument (most readable)
print(f'2000 m = {depth_convert(2000, unit="km"):.3f} km   (keyword arg)')
````

### 3c. Docstrings -- Documenting Your Functions

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A **docstring** is a string on the first line *inside* a function that explains what it does.
Write one in every function you create -- it appears when you call `help(function_name)`.

```python
def my_function(param):
    """
    One-line summary of what the function does.

    param  : what the input represents and its units / type
    Returns: what the output is
    """
    ...
```

**Note :**
We write/document code for other developers and our future self to easily understand it.
</div>

````python
def porosity_to_percent(phi):
    """
    Convert porosity from fraction to percentage.

    phi     : porosity as a fraction (0.0 to 1.0)
    Returns : porosity as a percentage (0.0 to 100.0)
    """
    return phi * 100.0

# Access the docstring interactively
help(porosity_to_percent)

# Use the function
print(f'Porosity 0.28 = {porosity_to_percent(0.28):.1f}%')
print(f'Porosity 0.32 = {porosity_to_percent(0.32):.1f}%')
````

### 3d. Returning Multiple Values

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A function can return **more than one value** by separating them with a comma.
Python packs them into a tuple; you can unpack them on the calling side.

```python
def stats(data):
    return min(data), max(data), sum(data)/len(data)

lo, hi, avg = stats(my_list)   # unpack three values at once
```

</div>

````python
def log_stats(values):
    """
    Return summary statistics for a log data list.

    values  : list of numeric readings
    Returns : (minimum, maximum, mean) as a tuple
    """
    lo   = min(values)
    hi   = max(values)
    mean = sum(values) / len(values)
    return lo, hi, mean

# Unpack three return values in one line
gr_min, gr_max, gr_mean = log_stats(gr_log)
print(f'GR  -- Min: {gr_min}    Max: {gr_max}    Mean: {gr_mean:.1f} API')

nphi_min, nphi_max, nphi_mean = log_stats(nphi_log)
print(f'NPHI -- Min: {nphi_min:.2f}  Max: {nphi_max:.2f}  Mean: {nphi_mean:.2f} V/V')
````

### 3e. Functions Calling Other Functions

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A function can call another function inside it.
This lets you build complex behaviour from small, well-tested building blocks.

```python
def step_one(x):
    return x * 2

def step_two(x):
    result = step_one(x)   # call step_one inside step_two
    return result + 10

step_two(5)   # returns 20
```

</div>

````python
# classify_gr() is already defined above.
# Write a richer function that calls it internally.

def describe_interval(depth, gr_val, nphi_val):
    """
    Produce a one-line description of a log interval.

    depth    : depth in metres
    gr_val   : GR reading in API units
    nphi_val : NPHI reading as a fraction
    Returns  : formatted string
    """
    litho = classify_gr(gr_val)   # call the existing function
    return f'Depth {depth:.1f} m | GR {gr_val:>3} API | NPHI {nphi_val:.2f} | {litho}'

# Test on individual intervals
print(describe_interval(depth_log[4], gr_log[4], nphi_log[4]))   # Good sand
print(describe_interval(depth_log[0], gr_log[0], nphi_log[0]))   # Shale
print(describe_interval(depth_log[7], gr_log[7], nphi_log[7]))   # Marginal sand

# Apply across the full log
print()
print('--- Full Log Description ---')
for d, g, n in zip(depth_log, gr_log, nphi_log):
    print(describe_interval(d, g, n))
````

## 4. Lambda Functions -- Deeper Dive

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

In **Lesson 4** we introduced lambda as a one-line shorthand for `def`:

```python
# Named function
def to_feet(m):
    return m * 3.28084

# Same as a lambda
to_feet = lambda m: m * 3.28084

# Conditional lambda
classify = lambda gr: "Sand" if gr < 75 else "Shale"
```

A lambda can only hold **one expression** -- no multiple lines, no assignments.
For anything more complex, use `def`.

Today we extend to: multi-condition lambdas, lambda as a sort key,
and combining lambda with loops.

</div>

````python
# Multi-condition lambda using a nested ternary expression
classify_gr_lam = lambda gr: (
    'Good sand'      if gr < 50  else
    ('Marginal sand'  if gr < 75  else
    'Shale')
)

# Verify it matches classify_gr() defined earlier
all_match = all(classify_gr_lam(g) == classify_gr(g) for g in gr_log)
print(f'Lambda matches classify_gr() on all readings: {all_match}')

# Preview first 8 results using zip + slice (no range needed)
print()
print(f'{"Depth":>8} | {"GR":>5} | {"Classification"}')
print('-' * 38)
for d, g in zip(depth_log[:8], gr_log[:8]):
    print(f'{d:>8.1f} | {g:>5} | {classify_gr_lam(g)}')
````

### 4a. Lambda as a Sort Key

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

`sorted()` can sort a list of dictionaries by any field using a `key` function.
A lambda is the ideal compact way to define that key.

```python
sorted_list = sorted(my_list, key=lambda item: item["field"])
```

- `key=` tells `sorted()` what value to compare.
- Add `reverse=True` to sort in descending order.
- Works on any list of complex objects -- dictionaries, tuples, custom classes.

</div>

````python
# List of wells with mean petrophysical properties
wells = [
    {'name': 'Well_A', 'gr_mean': 62.5,  'nphi_mean': 0.24},
    {'name': 'Well_B', 'gr_mean': 115.2, 'nphi_mean': 0.10},
    {'name': 'Well_C', 'gr_mean': 45.8,  'nphi_mean': 0.30},
    {'name': 'Well_D', 'gr_mean': 78.1,  'nphi_mean': 0.18},
]

# Rank by GR (lowest = cleanest sand first)
by_gr = sorted(wells, key=lambda w: w['gr_mean'])
print('Wells ranked by GR (cleanest first):')
for w in by_gr:
    print(f"  {w['name']}  GR mean = {w['gr_mean']:.1f} API")

print()

# Rank by NPHI (highest = most porous first)
by_nphi = sorted(wells, key=lambda w: w['nphi_mean'], reverse=True)
print('Wells ranked by porosity (most porous first):')
for w in by_nphi:
    print(f"  {w['name']}  NPHI mean = {w['nphi_mean']:.2f} V/V")
````

### 4b. When to Use lambda vs def

| Use `def` when | Use `lambda` when |
|----------------|-------------------|
| Logic spans more than one expression | Logic fits cleanly in one line |
| You need a docstring | No documentation needed |
| The function is called from many places | Used once (e.g. as a sort key) |
| The code is clearer written out in full | Inline is cleaner |

> **Rule of thumb:** if the lambda is hard to read at a glance, write a `def` instead.
> Readability matters more than brevity.

````python
# ENGINEERING SCENARIO: Automated Log Screener
# A while loop scans the GR log top to bottom.
# A function classifies each interval.
# A lambda calculates a quick Reservoir Quality Index (RQI).
# The screener stops at the first high-quality reservoir.

def classify_gr(gr):
    """Classify a GR reading into lithology."""
    if gr < 50:   return 'Good sand'
    elif gr < 75: return 'Marginal sand'
    else:         return 'Shale'

# Lambda: Reservoir Quality Index (0.0 to 1.0)
# High NPHI + low GR = better quality
rqi = lambda gr, nphi: round((1 - gr / 160.0) * nphi, 3)

print('--- Automated Log Screener ---')
print(f'{"Depth":>8} | {"GR":>5} | {"NPHI":>6} | {"Lithology":<15} | {"RQI":>5}')
print('-' * 56)

i = 0
target_found = False

while i < len(gr_log):
    d     = depth_log[i]
    g     = gr_log[i]
    n     = nphi_log[i]
    litho = classify_gr(g)
    score = rqi(g, n)

    print(f'{d:>8.1f} | {g:>5} | {n:>6.2f} | {litho:<15} | {score:>5.3f}')

    if litho == 'Good sand' and n > 0.28:
        print(f'\n>>> Target reservoir found at {d:.1f} m (RQI = {score}). Stopping scan.')
        target_found = True
        break

    i = i + 1

if not target_found:
    print('No high-quality reservoir found in the scanned interval.')
````

## 7. Exception Handling

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

Real programs fail sometimes: a user enters the wrong value, a list is empty, a file is missing, or a calculation divides by zero.
**Exception handling** helps your program respond safely instead of crashing without explanation.

```python
try:
    # code that might fail
except SomeError:
    # what to do if that error happens
else:
    # runs only when no error happens
finally:
    # always runs, whether there was an error or not
```

| Tool | Purpose |
|------|---------|
| `try` | Put risky code here |
| `except` | Handle a specific error |
| `else` | Run only if the `try` block succeeds |
| `finally` | Clean up or print a closing message |
| `raise` | Create your own clear error message |

</div>


````python
# Basic example: handle division by zero safely

def safe_divide(a, b):
    """
    Divide a by b safely.

    a       : numerator
    b       : denominator
    Returns : division result, or None if division is not possible
    """
    try:
        result = a / b
    except ZeroDivisionError:
        print('Cannot divide by zero.')
        return None
    else:
        return result

print(safe_divide(10, 2))
print(safe_divide(10, 0))
````

### 7a. Handling Bad User Input

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

Many beginner errors happen when input has the wrong type.
For example, `float("abc")` fails because Python cannot convert text letters into a number.

Use exception handling when an operation is valid in principle, but might fail because of the data.

Common examples:

| Error | Meaning |
|-------|---------|
| `ValueError` | Correct type, but invalid value, e.g. `float("abc")` |
| `TypeError` | Wrong type, e.g. adding a string to an integer |
| `IndexError` | List index is outside the valid range |
| `KeyError` | Dictionary key does not exist |
| `FileNotFoundError` | Requested file path does not exist |

</div>


````python
# Convert text input to a number safely

def parse_porosity(value):
    """
    Convert a value to porosity fraction.

    value   : input value, usually a string or number
    Returns : float porosity if valid, otherwise None
    """
    try:
        phi = float(value)
    except ValueError:
        print(f'Invalid porosity value: {value!r}')
        return None

    if phi < 0 or phi > 1:
        print('Porosity must be between 0 and 1.')
        return None

    return phi

print(parse_porosity('0.28'))
print(parse_porosity('bad-data'))
print(parse_porosity('1.5'))
````

### 7b. Raising Your Own Exceptions

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

Sometimes your function should reject bad input immediately.
Use `raise` when continuing would produce a misleading result.

Good error messages should explain:

1. what went wrong,
2. what value caused the problem,
3. what valid values are expected.

</div>


````python
# Engineering-style example: validate log data before calculating statistics

def validated_log_stats(values, log_name='log'):
    """
    Return minimum, maximum, and mean after validating a log list.

    values   : list of numeric readings
    log_name : name used in error messages
    Returns  : (minimum, maximum, mean)
    """
    if len(values) == 0:
        raise ValueError(f'{log_name} cannot be empty.')

    for value in values:
        if not isinstance(value, (int, float)):
            raise TypeError(f'{log_name} contains a non-numeric value: {value!r}')

    lo = min(values)
    hi = max(values)
    mean = sum(values) / len(values)
    return lo, hi, mean

# Valid data
print(validated_log_stats(gr_log, log_name='GR log'))

# Invalid data handled safely so the notebook continues running
try:
    print(validated_log_stats([], log_name='Empty log'))
except ValueError as error:
    print(f'Validation error: {error}')

try:
    print(validated_log_stats([10, 20, 'bad'], log_name='Test log'))
except TypeError as error:
    print(f'Validation error: {error}')
````

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

The `finally` block runs whether the operation succeeds or fails.
Use it for cleanup, closing files, releasing resources, or printing a final status message.

In beginner notebooks, `finally` is mostly useful for seeing the flow of execution clearly.

</div>


````python
# Demonstrate try / except / else / finally flow

def read_depth_at_index(index):
    """
    Return a depth value from depth_log using safe index access.
    """
    try:
        value = depth_log[index]
    except IndexError:
        print(f'Index {index} is outside the available depth range.')
        return None
    else:
        print(f'Depth at index {index}: {value:.1f} m')
        return value
    finally:
        print('Depth lookup attempt completed.
')

read_depth_at_index(3)
read_depth_at_index(100)
````

## 8. Variable Scope

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

**Variable scope** means the part of your program where a variable can be used.
Understanding scope helps you avoid confusing bugs, especially when functions use variables with the same name.

Python commonly uses these scopes:

| Scope | Meaning | Example |
|-------|---------|---------|
| Local | Created inside a function | `result` inside `def calculate():` |
| Global | Created outside all functions | `gr_log` in this notebook |
| Built-in | Provided by Python | `len`, `sum`, `print` |
| Enclosing | From an outer function around an inner function | nested functions |

The most important beginner rule:
**functions should usually receive inputs as parameters and return outputs instead of changing global variables directly.**

</div>


### 8a. Local Variables

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A variable created inside a function is **local** to that function.
It exists only while the function is running.

This protects your program because temporary variables inside a function do not accidentally overwrite variables outside the function.

</div>


````python
# Local variable example

def calculate_interval_quality(gr, nphi):
    quality_score = (1 - gr / 160.0) * nphi
    return quality_score

score = calculate_interval_quality(45, 0.32)
print(f'Quality score: {score:.3f}')

# quality_score is local to the function, so it is not available here.
try:
    print(quality_score)
except NameError as error:
    print(f'NameError: {error}')
````

### 8b. Reading Global Variables

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A function can read a global variable if it only needs to look at the value.
However, reading too many global variables makes functions harder to reuse and test.

Better design: pass important values as parameters.

</div>


````python
# A global threshold used by the function
global_gr_cutoff = 75

def is_sand_using_global(gr):
    return gr < global_gr_cutoff

print(is_sand_using_global(62))
print(is_sand_using_global(115))

# Better version: pass the cutoff as a parameter
def is_sand(gr, cutoff=75):
    return gr < cutoff

print(is_sand(62))
print(is_sand(115, cutoff=100))
````

### 8c. Modifying Global Variables

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

If a function reassigns a global variable, Python requires the `global` keyword.
Use this carefully.

```python
count = 0

def increment():
    global count
    count = count + 1
```

For clean code, prefer returning a new value and assigning it outside the function.

</div>


````python
# Modifying a global variable with global
sand_count = 0

def count_sand_interval(gr):
    global sand_count
    if gr < 75:
        sand_count = sand_count + 1

for g in gr_log:
    count_sand_interval(g)

print(f'Sand intervals counted with global variable: {sand_count}')

# Cleaner version: return the count instead of modifying global state
def count_sand_intervals(gr_values, cutoff=75):
    count = 0
    for gr in gr_values:
        if gr < cutoff:
            count = count + 1
    return count

clean_count = count_sand_intervals(gr_log)
print(f'Sand intervals counted with return value: {clean_count}')
````

### 8d. Mutable Global Objects

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

Lists and dictionaries are **mutable**, which means their contents can be changed.
A function can mutate a global list or dictionary without using `global`.

This is allowed, but it can be risky because the function changes data outside itself.

</div>


````python
# Mutating a global list does not require global
flags = []

def add_lithology_flag(gr):
    if gr < 75:
        flags.append('Sand')
    else:
        flags.append('Shale')

for g in gr_log[:5]:
    add_lithology_flag(g)

print(flags)

# Better version: create and return a new list
def build_lithology_flags(gr_values, cutoff=75):
    result = []
    for gr in gr_values:
        if gr < cutoff:
            result.append('Sand')
        else:
            result.append('Shale')
    return result

new_flags = build_lithology_flags(gr_log[:5])
print(new_flags)
````

### 8e. Enclosing Scope and `nonlocal`

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

An inner function can access variables from the outer function.
If the inner function must reassign an outer variable, use `nonlocal`.

You will not need `nonlocal` often as a beginner, but it explains how nested functions remember values.

</div>


````python
# Enclosing scope example

def make_counter():
    count = 0

    def increment():
        nonlocal count
        count = count + 1
        return count

    return increment

counter = make_counter()
print(counter())
print(counter())
print(counter())
````

| Mutable globals | `my_list.append(x)` | Change list/dict contents, but use carefully |
| Enclosing scope | nested functions | Inner function reads outer function variables |
| `nonlocal` | `nonlocal count` | Reassign a variable from an outer function |

### Best Practices

- Catch specific exceptions instead of using a bare `except`.
- Validate function inputs before doing important calculations.
- Use `raise` when invalid input should stop the function.
- Prefer function parameters and return values over global variables.
- Avoid changing global lists or dictionaries unless you clearly intend to.
- Keep functions small, predictable, and easy to test.

## 1. Why This Lesson, Right Now

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Bridge Lesson 1 got your notebooks onto GitHub -- version-controlled and shareable, but still solo. Every
`git commit` you've made so far has gone straight onto `main`, one person at a time, with nobody else
touching the same files at the same time.

That's not how real subsurface software or ML work happens. A reservoir engineering team shares a
history-matching codebase. A geoscience group shares a set of well-log processing notebooks. Everyone
edits the same files, often in the same week -- and if everyone just commits straight to `main`, two
people's work eventually collides.

Today's lesson is the two ideas that make shared repositories work: **branches**, and the **pull request**.
By the end, you'll cause a real merge conflict on purpose, fix it the same way a professional team does,
and know exactly what happens when you click the green "Merge pull request" button on GitHub.

</div>

## 2. The Problem With Everyone Committing to Main

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Imagine two engineers, Alex and Sam, both improving the same shared `porosity_helpers.py` file, both
committing straight to `main`:

- Alex starts editing a function on Monday morning. It's half-finished by lunch -- syntax errors, a
  `print()` left in for debugging.
- Sam pulls `main` at noon to add her own feature. She now has Alex's broken, half-finished code mixed
  into her own working copy, with no way to easily tell which lines are "done" and which aren't.
- If Alex pushes again before finishing, everyone downstream -- including anyone re-running the
  notebooks in this course from GitHub -- can end up with broken code on `main`, the one place everyone
  expects to always work.

**Branches** solve this: each person works on their own copy of the file, isolated from everyone else's
in-progress changes, and `main` only receives finished, reviewed work.

</div>

## 3. Branches: Your Own Sandbox Inside the Repo

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

A **branch** is a movable pointer to a commit. `main` is just a branch -- there is nothing structurally
special about it except convention. When you create a new branch, you get an independent line of commits
that starts from wherever you branched off, and changes on it do not affect `main` (or any other branch)
until you deliberately **merge** it back in.

```
main:      A---B---C-------------F      <- stays stable, only finished work lands here
                    \           /
feature:             D---E-----'        <- your in-progress work, isolated until merge
```

The everyday commands:

| Command | What it does |
|---|---|
| `git branch` | List branches; shows which one you're currently on |
| `git checkout -b <name>` | Create a new branch AND switch to it, in one step |
| `git checkout <name>` | Switch to an existing branch |
| `git merge <name>` | Bring another branch's commits into the branch you're currently on |
| `git branch -d <name>` | Delete a branch (once it's safely merged) |

</div>

## 4. Setting Up Our Team Simulation

<div style="background: #fff8e1; border-left: 5px solid #e0a800; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

To practice real collaboration without needing two separate GitHub accounts, we'll simulate GitHub with
a **bare repository** on this computer -- a repo with no working files of its own, only history, exactly
what GitHub itself hosts. Two folders, `alex_workspace` and `sam_workspace`, will each `git clone` it,
standing in for two different people's computers.

Every command from here on works **identically** against a real GitHub remote -- only the URL changes,
from a local folder path to something like `https://github.com/your-username/repo.git`.

</div>

````python
import os, shutil

demo_root = os.path.expanduser("~/git_team_demo")
if os.path.exists(demo_root):
    shutil.rmtree(demo_root)
os.makedirs(demo_root)
os.chdir(demo_root)
print(f"Working in: {os.getcwd()}")

# The "GitHub" stand-in: a bare repository (history only, no working files)
!git init --bare shared_remote.git -q
!cd shared_remote.git && git symbolic-ref HEAD refs/heads/main
print("Bare 'remote' repository created.")
````

````python
# Alex clones the (currently empty) remote and creates the first file
!git clone shared_remote.git alex_workspace -q

os.chdir(os.path.join(demo_root, "alex_workspace"))
# Local (not --global) identity -- this is what makes commits from this folder show as
# "Alex" specifically, separate from whatever --global identity is set on this machine.
!git config user.name "Alex"
!git config user.email "alex@example.com"
!git config commit.gpgsign false
!git config push.negotiate false

lines = [
    "def porosity_from_density(rho_matrix, rho_fluid, rho_log):",
    "    # Density-porosity equation.",
    "    return (rho_matrix - rho_log) / (rho_matrix - rho_fluid)",
    "",
]
with open("porosity_helpers.py", "w") as f:
    f.write("\n".join(lines))

!git add porosity_helpers.py
!git commit -m "Add porosity_from_density helper" -q
!git push -u origin main -q
print("--- Alex's initial commit, pushed to the shared remote ---")
!git log --oneline
````

`git config user.name` / `user.email`, run **without** `--global`, sets the identity for just this one
repository. Real teammates each have their own `--global` identity on their own machine, of course -- we
use local config here purely so this single notebook can show two distinct authors.

````python
# Sam clones the SAME remote -- she now has Alex's initial commit too
os.chdir(demo_root)
!git clone shared_remote.git sam_workspace -q

os.chdir(os.path.join(demo_root, "sam_workspace"))
!git config user.name "Sam"
!git config user.email "sam@example.com"
!git config commit.gpgsign false
!git config push.negotiate false

print("--- Sam's log immediately after cloning ---")
!git log --oneline
````

## 5. Contributor Alex: Branch, Commit, Push

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Alex wants to improve the one-line explanation of what the function does. Instead of editing `main`
directly, Alex creates a feature branch first -- the single habit that prevents Section 2's problem.

</div>

````python
os.chdir(os.path.join(demo_root, "alex_workspace"))
!git checkout -b feature/clearer-comment -q

content = open("porosity_helpers.py").read()
content = content.replace(
    "# Density-porosity equation.",
    "# Compute porosity from a density log using the standard density-porosity equation.",
)
with open("porosity_helpers.py", "w") as f:
    f.write(content)

!git add porosity_helpers.py
!git commit -m "Clarify explanatory comment on porosity_from_density" -q
!git push -u origin feature/clearer-comment -q
print("--- Alex's branch, pushed to the shared remote ---")
!git log --oneline --all
````

`main` on the remote is untouched -- `feature/clearer-comment` is a separate branch sitting alongside
it. Pushing a branch (instead of `main`) is exactly how you'd share work-in-progress on GitHub without
touching the stable version everyone else relies on.

## 6. Contributor Sam: Working in Parallel

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Meanwhile -- genuinely at the same time, this is the point -- Sam has her own idea for the same comment:
document the units. She branches off her own copy of `main`, which does **not** yet include Alex's change
(she cloned before he pushed his branch, and `main` itself hasn't moved).

</div>

````python
os.chdir(os.path.join(demo_root, "sam_workspace"))
!git checkout -b feature/add-units-comment -q

content = open("porosity_helpers.py").read()
content = content.replace(
    "# Density-porosity equation.",
    "# Porosity (fraction) from bulk density log, matrix density, and fluid density (all g/cc).",
)
with open("porosity_helpers.py", "w") as f:
    f.write(content)

!git add porosity_helpers.py
!git commit -m "Document units in porosity_from_density comment" -q
!git push -u origin feature/add-units-comment -q
print("--- Sam's branch, pushed to the shared remote ---")
!git log --oneline --all
````

Two branches now sit on the shared remote, both starting from the same original commit, both editing
the **exact same line** of the **exact same file**, differently. Neither Alex nor Sam has done anything
wrong -- this is completely ordinary parallel work. What happens next is the entire reason Section 8
exists.

## 7. Merging Without Conflict: Alex's Branch Goes First

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

On a real project, this is the moment Alex would open a **pull request** on GitHub and, after review
(Section 10-11), click **Merge**. Locally, that merge is just `git merge`. Since `main` hasn't changed
since Alex branched off it, this merge is completely clean.

</div>

````python
os.chdir(os.path.join(demo_root, "alex_workspace"))
!git checkout main -q
!git merge feature/clearer-comment -m "Merge feature/clearer-comment into main" -q
!git push origin main -q
print("--- main, after Alex's merge ---")
print(open("porosity_helpers.py").read())
````

## 8. Merge Conflicts: Causing One on Purpose, and Fixing It

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Sam is still on her branch, still holding the OLD comment as her starting point. Before she can merge her
branch into `main`, best practice is to bring `main`'s latest changes into her branch first and resolve
any overlap herself -- that's exactly what she's about to hit, since Alex's merged change touched the
same line she did.

</div>

````python
os.chdir(os.path.join(demo_root, "sam_workspace"))
!git checkout main -q
!git pull origin main -q
print("--- Sam's main, after pulling Alex's merged change ---")
!git log --oneline

!git checkout feature/add-units-comment -q
````

````python
import subprocess
result = subprocess.run(
    ["git", "merge", "main", "-m", "Merge main into feature/add-units-comment"],
    capture_output=True, text=True
)
print(result.stdout)
print(result.stderr)
print(f"merge exit code: {result.returncode}  (non-zero means: conflict, stop and look)")
````

`git status` is, once again, the first thing to run whenever something looks wrong -- it names exactly
which file(s) are conflicted and reminds you what to do next.

````python
!git status
````

````python
print(open("porosity_helpers.py").read())
````

<div style="background: #fff8e1; border-left: 5px solid #e0a800; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

### Reading Conflict Markers

Git couldn't decide which version you wanted, so it left **both**, wrapped in markers, and is waiting for
a human decision:

```
<<<<<<< HEAD
    (your current branch's version)
=======
    (the version being merged in)
>>>>>>> main
```

`HEAD` here means "the branch you're currently on" (Sam's `feature/add-units-comment`); `main` is the
branch being merged in. Resolving a conflict means **editing the file by hand** -- deleting the markers
and keeping, combining, or rewriting the content until only the version you actually want remains.

</div>

````python
# Resolve by hand: combine both engineers' improvements into one clearer comment
lines = [
    "def porosity_from_density(rho_matrix, rho_fluid, rho_log):",
    "    # Compute porosity (fraction) from a density log using the standard density-porosity",
    "    # equation. rho_matrix, rho_fluid, rho_log are all in g/cc.",
    "    return (rho_matrix - rho_log) / (rho_matrix - rho_fluid)",
    "",
]
with open("porosity_helpers.py", "w") as f:
    f.write("\n".join(lines))

print("--- resolved file (no more conflict markers) ---")
print(open("porosity_helpers.py").read())
!git status
````

````python
!git add porosity_helpers.py
!git commit -m "Merge main into feature/add-units-comment, resolve comment conflict" -q
!git push origin feature/add-units-comment -q
print("--- conflict resolved and pushed ---")
!git log --oneline --graph --all
````

A conflict is not an error you did something wrong -- it's Git correctly refusing to silently guess
which of two real, deliberate changes should win. Once resolved, Sam's branch can merge into `main` just
like Alex's did.

````python
!git checkout main -q
!git merge feature/add-units-comment -m "Merge feature/add-units-comment into main" -q
!git push origin main -q
print("--- final main ---")
print(open("porosity_helpers.py").read())
print("--- full project history ---")
!git log --oneline --graph --all
````

Notice there's no extra merge commit for this last step -- because Sam's branch already contained
`main`'s history plus her fix (from resolving the conflict), merging it back is a clean **fast-forward**.
Both engineers' work, and the record of exactly how the conflict was resolved, is now on `main`
permanently.

## 9. `git fetch` vs. `git pull`

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Both commands ask GitHub "what's new?" -- they differ in what happens next:

| Command | What it does |
|---|---|
| `git fetch` | Downloads new commits from the remote, but does **not** touch your working files. Safe to run any time -- lets you look before you leap with `git log origin/main` or `git diff main origin/main`. |
| `git pull` | Runs `git fetch`, then immediately `git merge` (or `rebase`, depending on configuration) into your current branch. What Section 8 used to bring `main`'s changes onto Sam's branch. |

A reasonable habit: `git fetch` freely and often just to see what teammates have pushed; `git pull` when
you're actually ready to bring those changes into the branch you're working on.

</div>

## 10. Pull Requests on Real GitHub

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Everything above happened with plain `git merge`, run locally. On GitHub, the same idea has a name and a
UI: a **Pull Request (PR)** -- literally a request to pull one branch into another, with a review step in
between. This can't be automated from inside this notebook (it needs your real GitHub login), so here is
exactly what it looks like:

1. Push a branch to GitHub -- `git push -u origin <branch-name>` -- exactly like Sections 5-6.
2. Go to your repository on **github.com**. GitHub usually shows a banner: *"`<branch-name>` had recent
   pushes -- Compare & pull request"*. Click it (or use the **Pull requests** tab -> **New pull request**).
3. Pick the **base** branch (usually `main`, what you want to merge INTO) and the **compare** branch
   (your feature branch, what you want to merge FROM).
4. Write a title and description -- what changed, and why. GitHub shows a **diff**: every added line in
   green, every removed line in red, file by file.
5. Click **Create pull request**. Anyone with access (a teammate, or just yourself) can now comment,
   request changes, or approve (Section 11).
6. Once satisfied, click **Merge pull request**. GitHub performs the exact same merge Sections 7-8 did by
   hand -- including stopping to ask you to resolve conflicts in the browser, if there are any.
7. Delete the branch (GitHub offers a button right there) -- its commits are safely part of `main`'s
   history now, so the branch itself has done its job.

**Why open a PR even working completely alone?** A PR is a checkpoint: it shows you the full diff before
it's permanent, it's a place to leave yourself a note about *why* you made a change, and it's the exact
workflow you'll already know the moment you do join a team.

</div>

## 11. Code Review Basics

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

A pull request's real value on a team is the review step before merging. On GitHub, a reviewer opens the
PR's **Files changed** tab and can:

- **Comment** on any specific line -- a question, a suggestion, or just a note.
- **Request changes** -- blocks merging until the author addresses the feedback (and pushes new commits,
  which automatically appear in the same PR).
- **Approve** -- signals the code is ready to merge.

Many teams configure a **branch protection rule** on `main`: at least one approving review is required
before the **Merge** button is even clickable, no matter who wrote the code. For a shared reservoir
simulation or ML codebase, this single setting is often what actually prevents Section 2's problem in
practice -- not good intentions, a rule the platform enforces.

When merging, GitHub usually offers a choice: a plain **merge commit** (what Sections 7-8 did -- keeps
every individual commit from the branch), **squash and merge** (combines the whole branch into one tidy
commit on `main`), or **rebase and merge** (replays the branch's commits on top of `main` with no merge
commit at all). Squash is a common default for small feature branches -- it keeps `main`'s history one
clean line, one PR per commit.

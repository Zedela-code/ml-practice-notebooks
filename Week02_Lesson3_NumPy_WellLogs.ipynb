## 1. Why NumPy? Lists vs Arrays

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

In **Lesson 2** we used Python lists to store sequences of values.  
Lists are flexible — but they were not designed for numerical computing.

**NumPy** (Numerical Python) provides the **array** — a data structure built specifically for fast, memory-efficient mathematical operations on large datasets.

In geoscience and petroleum engineering, we routinely work with:
- 🪨 Thousands of depth samples in a well log (one every 0.1 m)
- 🌊 Millions of seismic trace samples
- 📈 Years of daily production records

NumPy makes working with all of these fast and intuitive.

</div>

````python
# First, let's import NumPy — the standard alias is 'np'
import numpy as np

# ── Python list ──────────────────────────────────────────────────────────────
depths_list = [2000, 2050, 2100, 2150, 2200]

# To multiply every depth by 2 using a list, you need a loop (or list comprehension):
depths_doubled_list = [d * 2 for d in depths_list]
print("List doubled:  ", depths_doubled_list)

# ── NumPy array ──────────────────────────────────────────────────────────────
depths_array = np.array([2000, 2050, 2100, 2150, 2200])

# With NumPy, you just write the math directly — no loop needed:
depths_doubled_array = depths_array * 2
print("Array doubled: ", depths_doubled_array)
````

> **Key insight:** With a Python list, multiplying by 2 requires writing a loop.  
> With a NumPy array, you write `array * 2` and NumPy applies the operation to **every element at once**.  
> This is called **vectorization** — and it is dramatically faster on large datasets.

## 2. Creating NumPy Arrays

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

There are several ways to create NumPy arrays. The three most useful for engineering work are:

| Method | What it creates | Example |
|--------|----------------|---------|
| `np.array([...])` | Array from an existing list | `np.array([1, 2, 3])` |
| `np.arange(start, stop, step)` | Evenly spaced values (like `range`) | `np.arange(0, 10, 2)` |
| `np.linspace(start, stop, n)` | `n` evenly spaced values between start and stop | `np.linspace(0, 1, 5)` |

</div>

````python
import numpy as np

# Method 1: Convert a Python list to a NumPy array
gr_values = np.array([45, 62, 88, 120, 95, 55, 40])  # Gamma Ray readings (API)

print("Array:  ", gr_values)
print("Type:   ", type(gr_values))
print("Shape:  ", gr_values.shape)     # how many elements (rows, cols)
print("Dtype:  ", gr_values.dtype)     # data type of elements
````

````python
import numpy as np

# Method 2: np.arange — depth log from 2000 m to 2100 m, sampled every 0.5 m
depth = np.arange(2000, 2100.5, 0.5)   # start, stop (exclusive), step

print(f"Depth array (first 5): {depth[:5]} m")
print(f"Depth array (last 5):  {depth[-5:]} m")
print(f"Total samples:         {len(depth)}")
print(f"Min depth: {depth.min()} m   Max depth: {depth.max()} m")
````

````python
import numpy as np

# Method 3: np.linspace — create exactly 6 evenly spaced pressure values
pressure_gradient = np.linspace(1000, 4500, 6)   # 6 values from 1000 to 4500 psi

print("Pressure gradient (psi):", pressure_gradient)

# Note the difference from arange:
#   arange: you specify the STEP SIZE
#   linspace: you specify the NUMBER OF POINTS
````

## 3. Vectorized Operations

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

**Vectorization** means writing one mathematical expression that applies to an entire array automatically — no loops.

This is the single most important NumPy concept. It is what makes scientific Python code look clean and run fast.

</div>

````python
import numpy as np

porosity = np.array([0.12, 0.18, 0.22, 0.15, 0.30])   # fraction

# Arithmetic — applied to every element
porosity_percent = porosity * 100        # convert to percentage
print("Porosity (fraction):", porosity)
print("Porosity (%):       ", porosity_percent)

# More operations
print("\nMin porosity: ", porosity.min())
print("Max porosity: ", porosity.max())
print("Mean porosity:", porosity.mean())
````

### Array Comparisons — Boolean Masks

You can compare an array to a value and get a **boolean array** (True/False for each element).  
This is used to **filter data** — a technique you will use constantly in well-log analysis:

````python
import numpy as np

gr = np.array([25, 48, 130, 90, 35, 145, 60, 112, 28, 70])   # Gamma Ray (API)

# Which samples are above 75 API? (likely shale)
is_shale = gr > 75
print("GR values:   ", gr)
print("Is shale:    ", is_shale)

# Filter: keep only shale samples
shale_gr = gr[is_shale]
print("\nShale GR values:", shale_gr)
print(f"Shale count: {is_shale.sum()} of {len(gr)} samples")
````

> **What just happened?**  
> `gr > 75` creates a True/False array.  
> `gr[is_shale]` uses that mask to select only the elements where the condition is True.  
> This pattern — **create a mask, then filter** — is the foundation of all well-log facies analysis.

## 4. Essential NumPy Functions

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

NumPy has a rich set of built-in functions for numerical analysis. Here are the ones you will use most often in well-log work:

| Function | Description | Geoscience use |
|----------|-------------|----------------|
| `np.mean(a)` | Arithmetic mean | Average porosity of a zone |
| `np.std(a)` | Standard deviation | Variability in GR response |
| `np.median(a)` | Median value | Robust central tendency |
| `np.percentile(a, q)` | q-th percentile | Reservoir quality cutoffs |
| `np.where(cond, x, y)` | Conditional assignment | Lithology classification |
| `np.cumsum(a)` | Cumulative sum | Cumulative production |
| `np.diff(a)` | Differences between consecutive elements | Rate of change with depth |

</div>

````python
import numpy as np

# Simulated Gamma Ray log — 20 samples
gr = np.array([30, 45, 80, 120, 140, 95, 60, 35, 25, 50,
               130, 115, 88, 42, 38, 70, 145, 55, 28, 90])

print(f"Mean GR:        {np.mean(gr):.1f} API")
print(f"Median GR:      {np.median(gr):.1f} API")
print(f"Std deviation:  {np.std(gr):.1f} API")
print(f"Min GR:         {np.min(gr):.0f} API")
print(f"Max GR:         {np.max(gr):.0f} API")
print(f"10th percentile:{np.percentile(gr, 10):.1f} API  (cleaner sands are below this)")
print(f"90th percentile:{np.percentile(gr, 90):.1f} API  (shales are above this)")
````

### `np.where` — Classify Lithology in One Line

````python
import numpy as np

gr = np.array([30, 45, 80, 120, 140, 95, 60, 35, 25, 50,
               130, 115, 88, 42, 38, 70, 145, 55, 28, 90])

# Standard cutoff: GR < 75 API → Sand; GR >= 75 API → Shale
CUTOFF = 75
lithology = np.where(gr < CUTOFF, "Sand ", "Shale")

print("GR value  | Lithology")
print("-" * 22)
for g, lith in zip(gr, lithology):
    print(f"  {g:>3} API | {lith}")
````

### `np.cumsum` — Cumulative Production

In production engineering, cumulative production is one of the most important performance indicators:

````python
import numpy as np

# Monthly oil production (barrels) for 6 months
monthly_production = np.array([4500, 4200, 3900, 3600, 3400, 3200])

cumulative = np.cumsum(monthly_production)

print(f"{'Month':>6}  {'Monthly (bbl)':>14}  {'Cumulative (bbl)':>16}")
print("-" * 42)
for i, (m, c) in enumerate(zip(monthly_production, cumulative), 1):
    print(f"{i:>6}  {m:>14,}  {c:>16,}")
````

## 5. Simulating Well-Log Data

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Before we work with real LAS files (coming in a later lesson), we will **simulate** a well log using NumPy.

Simulated data lets us:
- Practice without needing a file
- Control the exact values for learning purposes
- Understand what real logs look like conceptually

We will create a simple **Gamma Ray (GR) log** — the most fundamental well log in formation evaluation.

**Gamma Ray** measures natural radioactivity of the formation:
- **Low GR (20–75 API)** → Sands and carbonates (reservoir rocks)
- **High GR (75–150+ API)** → Shales (non-reservoir)

</div>

````python
import numpy as np

# ── Simulate depth axis: 2000 m to 2500 m, sampled every 0.5 m ──────────────
depth = np.arange(2000, 2500.5, 0.5)  # 1001 depth samples

# ── Simulate a Gamma Ray log ─────────────────────────────────────────────────
# We use np.random.seed for reproducibility — same "random" numbers every run
np.random.seed(42)

# Base GR signal — starts at 80, varies with depth
gr_base = 80 + 30 * np.sin(np.linspace(0, 4 * np.pi, len(depth)))

# Add some random noise (natural variability in the formation)
noise = np.random.normal(loc=0, scale=10, size=len(depth))
gr = gr_base + noise

# Clip to realistic GR range (10–150 API)
gr = np.clip(gr, 10, 150)

print(f"Depth: {depth[0]:.0f} m to {depth[-1]:.0f} m  |  {len(depth)} samples")
print(f"GR log — Mean: {gr.mean():.1f}  Std: {gr.std():.1f}  Min: {gr.min():.1f}  Max: {gr.max():.1f} API")
````

### Simulating a Porosity Log

We also simulate a **Neutron Porosity (NPHI)** log. In real logs, porosity is inversely correlated with GR  
(sands have high porosity, shales have low porosity):

````python
import numpy as np

# Rebuild the depth and GR arrays
np.random.seed(42)
depth  = np.arange(2000, 2500.5, 0.5)
gr_base = 80 + 30 * np.sin(np.linspace(0, 4*np.pi, len(depth)))
gr     = np.clip(gr_base + np.random.normal(0, 10, len(depth)), 10, 150)

# Simulate NPHI: inversely correlated with GR, with noise
# GR ~ 80 → NPHI ~ 0.25; GR ~ 150 → NPHI ~ 0.05
nphi_base = 0.35 - (gr / 150) * 0.25
nphi      = np.clip(nphi_base + np.random.normal(0, 0.02, len(depth)), 0.02, 0.45)

print(f"NPHI — Mean: {nphi.mean():.3f}  Min: {nphi.min():.3f}  Max: {nphi.max():.3f}")

# Quick sanity check — count sand vs shale samples
sand_count  = (gr < 75).sum()
shale_count = (gr >= 75).sum()
print(f"\nSand samples (GR < 75):  {sand_count} ({sand_count/len(gr)*100:.0f}%)")
print(f"Shale samples (GR ≥ 75): {shale_count} ({shale_count/len(gr)*100:.0f}%)")
````

## 6. First Well-Log Plot with Matplotlib

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

**Matplotlib** is Python's standard plotting library. We use it to visualise our well logs — the same way a petrophysicist would view them in dedicated software.

> **Convention in well-log display:**  
> - Depth is plotted on the **vertical axis**, increasing **downward**  
> - Log values are plotted on the **horizontal axis**  
> - This mimics the physical reality of drilling a borehole

We will use `matplotlib.pyplot` — the standard alias is `plt`.

</div>

````python
import numpy as np
import matplotlib.pyplot as plt

# ── Rebuild the simulated log data ────────────────────────────────────────────
np.random.seed(42)
depth   = np.arange(2000, 2500.5, 0.5)
gr_base = 80 + 30 * np.sin(np.linspace(0, 4*np.pi, len(depth)))
gr      = np.clip(gr_base + np.random.normal(0, 10, len(depth)), 10, 150)
nphi_base = 0.35 - (gr / 150) * 0.25
nphi    = np.clip(nphi_base + np.random.normal(0, 0.02, len(depth)), 0.02, 0.45)

print("Data ready — running the plot cell next...")
````

````python
# ── Plot 1: Gamma Ray log (single track) ─────────────────────────────────────
fig, ax = plt.subplots(figsize=(4, 10))

ax.plot(gr, depth, color='#2d6a9f', linewidth=0.8, label='GR')
ax.axvline(x=75, color='#e07b39', linestyle='--', linewidth=1.2, label='Sand/Shale cutoff (75 API)')

# Fill sand zones (GR < 75) with yellow
ax.fill_betweenx(depth, gr, 75, where=(gr < 75), color='#f5e642', alpha=0.4, label='Sand')

# Axis formatting
ax.set_xlabel('Gamma Ray (API)', fontsize=11)
ax.set_ylabel('Depth (m)', fontsize=11)
ax.set_title('Simulated GR Log\nWell A-01', fontsize=12, fontweight='bold')
ax.invert_yaxis()   # depth increases downward
ax.set_xlim(0, 160)
ax.legend(loc='lower right', fontsize=9)
ax.grid(axis='x', linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
````

### Two-Track Display — GR and NPHI Side by Side

In practice, petrophysicists display multiple logs side-by-side in **tracks**. Let's plot GR and NPHI together:

````python
# ── Plot 2: Two-track log display ─────────────────────────────────────────────
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(7, 10), sharey=True)

# Track 1: Gamma Ray
ax1.plot(gr, depth, color='#2d6a9f', linewidth=0.8)
ax1.axvline(x=75, color='#e07b39', linestyle='--', linewidth=1.0)
ax1.fill_betweenx(depth, gr, 75, where=(gr < 75), color='#f5e642', alpha=0.35)
ax1.set_xlabel('Gamma Ray (API)', fontsize=10)
ax1.set_title('GR', fontsize=11, fontweight='bold')
ax1.set_xlim(0, 160)
ax1.invert_yaxis()
ax1.grid(axis='x', linestyle='--', alpha=0.3)

# Track 2: Neutron Porosity
ax2.plot(nphi, depth, color='#2ca87f', linewidth=0.8)
ax2.fill_betweenx(depth, 0, nphi, where=(gr < 75), color='#2ca87f', alpha=0.25)
ax2.set_xlabel('NPHI (fraction)', fontsize=10)
ax2.set_title('NPHI', fontsize=11, fontweight='bold')
ax2.set_xlim(0, 0.50)
ax2.grid(axis='x', linestyle='--', alpha=0.3)

fig.supylabel('Depth (m)', fontsize=11)
fig.suptitle('Simulated Well Log — Well A-01', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()
print("\nYellow bands = Sand zones (GR < 75 API)  |  Green fill = Porosity in sand zones")
````

> **Reading the plot:**  
> - Yellow bands on the GR track indicate **sand intervals** (GR below the 75 API cutoff).  
> - On the NPHI track, the green fill shows that porosity is **higher in the same sand intervals** — exactly what we expect physically.  
> - This is the core of petrophysical well-log analysis: correlating multiple logs to identify reservoir quality.

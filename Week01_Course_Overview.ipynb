## 2. What Is Machine Learning?

### The AI Landscape

````python
# This diagram was pre-generated for you — no need to understand this code yet.
from IPython.display import Image
Image('../assets/ai_landscape.png')
````

> **What the diagram shows:** AI is the broadest category — it includes any technique that mimics human-like reasoning. Machine Learning (ML) is a subset of AI that *learns patterns from data* rather than following hand-coded rules. Deep Learning is a subset of ML that uses layered networks loosely inspired by the brain. In this course we cover ML and Deep Learning.

### The Three Key Learning Paradigms

| Paradigm | How It Learns | Petroleum Example |
|----------|--------------|-------------------|
| **Supervised Learning** | Learns from labelled input-output pairs | Predicting permeability from well-log data |
| **Unsupervised Learning** | Finds hidden structure in unlabelled data | Automatic reservoir zonation by clustering |
| **Reinforcement Learning** | Learns by trial and reward | Drilling parameter optimisation |

### Classification vs. Regression

```
Is the output...
    ├── A category?   →  CLASSIFICATION   (e.g., sandstone / shale / carbonate)
    └── A number?     →  REGRESSION       (e.g., porosity = 18.4%)
```

### Goal
Get your environment running and load your first real dataset.

The next cell creates a **synthetic** (computer-simulated) well-log dataset — a table with 1,000 rows, one for each half-metre of depth in an imaginary well. Each row has four columns:

| Column | Full Name | What It Measures |
|--------|-----------|-----------------|
| `DEPTH` | Measured Depth | How far down the borehole (metres) |
| `GR` | Gamma Ray | Natural radioactivity — high GR = shale, low GR = sand |
| `RHOB` | Bulk Density | Mass per unit volume of rock (g/cm³) |
| `POROSITY` | Porosity | Fraction of the rock that is empty pore space (0 = solid rock, 0.38 = very porous) |

> 📌 **DataFrame:** In Python, a `DataFrame` (from the `pandas` library) is like a spreadsheet — rows are measurements, columns are variables. You will use DataFrames in every notebook in this course.

**Run the cell below.** You do not need to understand every line of the code — focus on reading the output table and statistics that appear.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ─── Synthetic well-log dataset ───────────────────────────────────────────────
# (In Week 3 we will load a real LAS file using LASIO)
np.random.seed(42)
depth = np.arange(2000, 2500, 0.5)   # measured depth in metres
n = len(depth)

# Simulate three petrophysical logs
GR      = 40 + 60 * np.abs(np.sin(depth / 30)) + np.random.normal(0, 5, n)   # Gamma Ray (API)
RHOB    = 2.1 + 0.4 * (1 - np.abs(np.sin(depth / 30))) + np.random.normal(0, 0.03, n)  # Density (g/cc)
POROSITY = np.clip(0.30 - 0.25 * (RHOB - 2.1) / 0.4 + np.random.normal(0, 0.02, n), 0.05, 0.38)

df = pd.DataFrame({
    'DEPTH':    depth,
    'GR':       GR.round(2),
    'RHOB':     RHOB.round(3),
    'POROSITY': POROSITY.round(3)
})

print("=== Well Log Dataset ===")
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"Depth range: {df.DEPTH.min()} – {df.DEPTH.max()} m")
print()
print(df.head(8).to_string(index=False))
print()
print("=== Basic Statistics ===")
print(df.describe().round(3).to_string())
````

````python
# ─── First petroleum plot: a 3-track well log display ─────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(11, 9), sharey=True)
fig.patch.set_facecolor('#f9fbff')

track_cfg = [
    ('GR',       'Gamma Ray (API)',      '#2d6a9f', (0, 150)),
    ('RHOB',     'Bulk Density (g/cc)',  '#e07b39', (1.9, 2.8)),
    ('POROSITY', 'Porosity (fraction)',  '#2ca87f', (0, 0.4)),
]

for ax, (col, label, color, xlim) in zip(axes, track_cfg):
    ax.plot(df[col], df['DEPTH'], color=color, linewidth=0.8, alpha=0.9)
    ax.set_xlabel(label, fontsize=10, fontweight='bold', color=color)
    ax.set_xlim(xlim)
    ax.set_facecolor('#f9fbff')
    ax.invert_yaxis()
    ax.grid(True, linestyle='--', alpha=0.35)
    ax.tick_params(labelsize=8)

axes[0].set_ylabel('Measured Depth (m)', fontsize=11, color='#1a3a5c')
fig.suptitle('Synthetic Well Log Display\
(Simulated — Real data from Week 3 onward)',
             fontsize=13, fontweight='bold', color='#1a3a5c', y=1.01)

plt.tight_layout()
plt.savefig('../assets/well_log_display.png', dpi=120, bbox_inches='tight', facecolor='#f9fbff')
plt.show()
print("Your first petroleum data visualisation — well done!")
````

> **What to notice:** Depth increases downward — the same way you'd see it in a real borehole report. The Gamma Ray (GR) track oscillates: high values indicate clay-rich shale, low values indicate cleaner sandstone. Notice that where RHOB (density) is high, POROSITY tends to be low — denser rock has less pore space. **This inverse relationship is exactly the kind of pattern ML will learn to detect automatically.**

## 10b. Useful Tools — Quick Reference

The assignment asks you to use three Python patterns. Here they are fully worked out — **run each cell to see the output**, then copy the same structure into your assignment.

### A: Computing Statistics

````python
# How to compute basic statistics for a single column
# Use df['COLUMN_NAME'] to select one column, then call a method on it.
print("─── Gamma Ray (GR) Statistics ───")
print(f"  Mean:  {df['GR'].mean():.2f}")    # average value
print(f"  Std:   {df['GR'].std():.2f}")     # spread — how much values vary
print(f"  Min:   {df['GR'].min():.2f}")
print(f"  Max:   {df['GR'].max():.2f}")
print(f"  Range: {df['GR'].max() - df['GR'].min():.2f}")   # max minus min
````

### B: Measuring Correlation

**Correlation** tells us how strongly two measurements are linked. The result is always between −1 and +1.

````python
# Compute the correlation between two columns.
# Correlation measures how strongly two measurements move together:
#   +1.0 = perfect positive relationship
#   -1.0 = perfect negative relationship
#    0.0 = no relationship
corr_value = df['RHOB'].corr(df['POROSITY'])
print(f"Correlation between RHOB and POROSITY: {corr_value:.3f}")
if corr_value < -0.5:
    print("→ Strong negative correlation: as density increases, porosity decreases.")
````

### C: Creating a Scatter Plot (Crossplot)

A **scatter plot** (also called a crossplot in geoscience) shows the relationship between two measurements. Each dot is one depth point. Run the cell below — you will use this same pattern in the assignment.

````python
import matplotlib.pyplot as plt

# A scatter plot shows the relationship between two measurements.
# Each dot represents one depth point in the well.
plt.figure(figsize=(6, 4))
plt.scatter(df['RHOB'], df['POROSITY'], color='steelblue', alpha=0.3, s=8)

plt.xlabel('Bulk Density RHOB (g/cc)')
plt.ylabel('Porosity (fraction)')
plt.title('RHOB vs Porosity — Crossplot')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print("Notice the trend: higher density → lower porosity.")
````

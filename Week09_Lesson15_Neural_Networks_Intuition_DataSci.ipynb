## 1. Where We Are: From Classical ML to Deep Learning

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Lessons 8-12 were all **classical ML**: Decision Trees, Random Forests, K-Means, PCA. Every one of those
models makes predictions using rules, distances, or votes that you can trace by hand.

Starting today, we move into **Deep Learning** -- models built from many simple units (neurons) connected
in layers, trained by gradually adjusting connection strengths until the network's predictions improve.

This is not a replacement for what you already know. A Random Forest is often the better tool for tabular
business data like ours (~200 cars, 10 features). Neural networks earn their keep on problems classical ML
struggles with -- images, text, and very large datasets -- which is where this course is headed over the
next few weeks. Today's goal is intuition: understand what a neural network *is* before asking what it is good for.

</div>

## 2. The Biological Inspiration

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

A biological neuron is simple in outline:

```
Dendrites (inputs)  -->  Cell Body (sums the signals)  -->  Axon (output, if the sum is strong enough)
```

It receives signals from many other neurons, each connection with its own strength (synapse). If the
combined, weighted signal is strong enough, the neuron "fires" and passes a signal onward.

An **artificial neuron** copies this structure with numbers instead of chemistry:

| Biological | Artificial |
|---|---|
| Dendrite input | A feature value (e.g. `engine_size`) |
| Synapse strength | A weight |
| Cell body summing signals | A weighted sum plus a bias |
| "Fires" or not | An activation function decides the output |

</div>


---
## 3. The Artificial Neuron

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

A single artificial neuron does exactly three things:

```
1. Multiply each input by its weight
2. Add them all up, plus a bias term
3. Pass the result through an activation function
```

$$z = (w_1 x_1 + w_2 x_2 + \dots + w_n x_n) + b \qquad \text{output} = f(z)$$

Let's compute one neuron by hand with three made-up scaled specification inputs, so the formula stops being abstract.

</div>


````python
import numpy as np

# Three toy inputs, standing in for scaled engine_size, horsepower, curb_weight values
x = np.array([0.9, 0.6, -0.4])
w = np.array([0.5, 0.4, -0.2])   # learned weights (made up for this demo)
b = -0.1                          # learned bias

z = np.dot(x, w) + b
output = 1 / (1 + np.exp(-z))     # sigmoid activation

print(f"Weighted sum + bias (z): {z:.3f}")
print(f"Neuron output (sigmoid): {output:.3f}")
````

> That is the entire computation inside **one** neuron. A real network chains hundreds of these together.
> The weights and bias are not made up in practice -- they start random and are learned automatically
> during training, the same way `RandomForestClassifier.fit()` learns tree splits automatically.


---
## 4. Activation Functions

<div style="background: #fff8e1; border-left: 5px solid #e0a800; padding: 14px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

If every neuron only computed a weighted sum, stacking layers would be pointless -- a chain of purely
linear steps collapses into one linear step, no more powerful than a Linear Regression.

The **activation function** injects non-linearity, which is what lets a network learn curved, complex
decision boundaries. Three common choices:

</div>


````python
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 200)
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)
relu = np.maximum(0, x)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(x, sigmoid, label="Sigmoid", color="#2d6a9f", linewidth=2)
ax.plot(x, tanh, label="Tanh", color="#2ca87f", linewidth=2)
ax.plot(x, relu, label="ReLU", color="#e07b39", linewidth=2)
ax.axhline(0, color="gray", linewidth=0.7)
ax.axvline(0, color="gray", linewidth=0.7)
ax.set_xlabel("z (weighted sum + bias)")
ax.set_ylabel("Activation output")
ax.set_title("Common Activation Functions")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

> **ReLU** (`max(0, z)`) is the default in most modern networks, including the one we build today --
> it is cheap to compute and trains faster than sigmoid or tanh in deep networks.

## 5. From One Neuron to a Network

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Neurons are arranged in **layers**. Our car pricing problem maps onto this structure directly:

```
INPUT LAYER          HIDDEN LAYER            OUTPUT LAYER
(10 specifications)  (8 neurons, ReLU)        (3 price categories)

wheel_base   --\                                --> Low
length       --  \                              --> Medium
width        ---- [ 8 neurons ] --------------->  --> High
...          --  /
highway_mpg  --/
```

Every input connects to every hidden neuron (with its own weight), and every hidden neuron connects to
every output. This is called a **Multi-Layer Perceptron (MLP)** -- the most basic kind of neural network,
and exactly what scikit-learn's `MLPClassifier` builds for us.

Training means: start with random weights, make a prediction, measure the error, and nudge every weight
slightly to reduce that error -- repeated thousands of times. That adjustment process is called
**backpropagation**. We treat it as a black box today; Lesson 14 opens it up with PyTorch.

</div>

## 6. Prepare and Scale the Data

Same cleaned `imports-85` data, the same ten features and `price_category` target used since Lesson 9.
Like K-Means and PCA in Lesson 11, neural networks learn through numerical weight updates, so features
must be **scaled** -- unscaled `curb_weight` (thousands) would dominate unscaled `compression_ratio` (single digits).


````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Load and clean data
df = pd.read_csv("../data/imports-85.data", header=None)
col_names = [
    "symboling", "normalized_losses", "make", "fuel_type", "aspiration",
    "num_of_doors", "body_style", "drive_wheels", "engine_location",
    "wheel_base", "length", "width", "height", "curb_weight",
    "engine_type", "num_of_cylinders", "engine_size", "fuel_system",
    "bore", "stroke", "compression_ratio", "horsepower", "peak_rpm",
    "city_mpg", "highway_mpg", "price"
]
df.columns = col_names
df = df.replace("?", np.nan)
numeric_cols = ["normalized_losses", "bore", "stroke", "horsepower", "peak_rpm", "price"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df = df.dropna(subset=["price"]).reset_index(drop=True)

def price_category(p):
    if p < 10000:
        return "Low"
    elif p < 20000:
        return "Medium"
    else:
        return "High"

df["price_category"] = df["price"].apply(price_category)

feature_cols = [
    "wheel_base", "length", "width", "height", "curb_weight",
    "engine_size", "compression_ratio", "horsepower",
    "city_mpg", "highway_mpg"
]
class_order = ["Low", "Medium", "High"]

df_model = df[feature_cols + ["price_category"]].dropna().reset_index(drop=True)
X = df_model[feature_cols].astype(float)
y = df_model["price_category"]
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Training rows: {len(X_train)}   Test rows: {len(X_test)}")
````

---
## 7. Build Your First Neural Network

One hidden layer of 8 neurons, ReLU activation -- the architecture sketched in Section 5.
`MLPClassifier` handles the weighted sums, activations, and backpropagation internally.


````python
mlp = MLPClassifier(
    hidden_layer_sizes=(8,),
    activation="relu",
    solver="adam",
    max_iter=2000,
    random_state=42
)
mlp.fit(X_train_scaled, y_train)

mlp_pred = mlp.predict(X_test_scaled)
mlp_acc = accuracy_score(y_test, mlp_pred)

print(f"Neural network test accuracy: {mlp_acc:.3f}")
print(f"Training stopped after {mlp.n_iter_} iterations")
print("\nClassification report:")
print(classification_report(y_test, mlp_pred, labels=class_order, target_names=class_order))
````

````python
# Watch the network learn: training loss should fall steadily
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(mlp.loss_curve_, color="#2d6a9f", linewidth=2)
ax.set_xlabel("Training Iteration")
ax.set_ylabel("Loss")
ax.set_title("Neural Network Training Curve")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
````

## 8. Neural Network vs Random Forest

<div style="background: #fff8e1; border-left: 5px solid #e0a800; padding: 14px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

This is the comparison every data scientist makes before choosing a model for a tabular problem like ours.
We refit the Lesson 10 baseline Random Forest settings here as a fair, like-for-like comparison.

</div>


````python
rf = RandomForestClassifier(
    n_estimators=100, max_depth=5, random_state=42, class_weight="balanced"
)
rf.fit(X_train, y_train)   # trees do not need scaled features
rf_acc = accuracy_score(y_test, rf.predict(X_test))

print(f"Random Forest test accuracy : {rf_acc:.3f}")
print(f"Neural Network test accuracy: {mlp_acc:.3f}")

fig, ax = plt.subplots(figsize=(5.5, 4))
ax.bar(["Random Forest", "Neural Network"], [rf_acc, mlp_acc], color=["#2ca87f", "#2d6a9f"])
ax.set_ylabel("Test Accuracy")
ax.set_ylim(0, 1)
ax.set_title("Model Comparison on the Same Test Set")
for i, acc in enumerate([rf_acc, mlp_acc]):
    ax.text(i, acc + 0.02, f"{acc:.3f}", ha="center")
plt.tight_layout()
plt.show()
````

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

### Reading the Comparison

- On small, tabular datasets like ours (~200 cars, 10 features), Random Forest usually matches or beats a
  neural network -- trees need less data and less tuning to reach a good result.
- Neural networks tend to pull ahead on much larger datasets, or on data with structure trees cannot
  exploit directly (images, text, sequences) -- exactly where this course is headed next.
- Neither model is "better" in general. Picking one is an engineering decision based on data size,
  data type, and how much you need to explain the result to a client (recall Lesson 12).

</div>

## 9. Visualizing the Decision Boundary

Ten features cannot be plotted directly, so we train small 2-feature versions of both models on
`curb_weight` and `engine_size` only -- the two strongest predictors from Lesson 10's feature importance --
purely to see the *shape* of boundary each model draws.


````python
from matplotlib.colors import ListedColormap

f1, f2 = "curb_weight", "engine_size"
X2 = df_model[[f1, f2]].values.astype(float)
y2 = df_model["price_category"].values
class_to_int = {c: i for i, c in enumerate(class_order)}
y2_int = np.array([class_to_int[c] for c in y2])

scaler2 = StandardScaler()
X2_scaled = scaler2.fit_transform(X2)

rf2 = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42, class_weight="balanced")
rf2.fit(X2, y2)

mlp2 = MLPClassifier(hidden_layer_sizes=(8,), activation="relu", max_iter=2000, random_state=42)
mlp2.fit(X2_scaled, y2)

print("Two lightweight 2-feature models trained for visualization only.")
````

````python
cmap = ListedColormap(["#2ca87f", "#e0a800", "#c0392b"])

x_min, x_max = X2[:, 0].min() - 200, X2[:, 0].max() + 200
y_min, y_max = X2[:, 1].min() - 10, X2[:, 1].max() + 10
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))
grid = np.c_[xx.ravel(), yy.ravel()]

rf_regions = rf2.predict(grid)
rf_regions = np.array([class_to_int[c] for c in rf_regions]).reshape(xx.shape)

grid_scaled = scaler2.transform(grid)
mlp_regions = mlp2.predict(grid_scaled)
mlp_regions = np.array([class_to_int[c] for c in mlp_regions]).reshape(xx.shape)

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
for ax, regions, title in zip(axes, [rf_regions, mlp_regions], ["Random Forest", "Neural Network"]):
    ax.contourf(xx, yy, regions, cmap=cmap, alpha=0.5, levels=np.arange(-0.5, 3.5, 1))
    ax.scatter(X2[:, 0], X2[:, 1], c=y2_int, cmap=cmap, s=16, edgecolor="white", linewidth=0.3)
    ax.set_xlabel(f1)
    ax.set_ylabel(f2)
    ax.set_title(f"{title} Decision Regions")

fig.suptitle("Same Two Features, Two Very Different Boundary Shapes", fontsize=12, y=1.02)
plt.tight_layout()
plt.show()
````

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

### What to Look For

- The **Random Forest** boundary is made of straight, axis-aligned steps -- a direct result of asking
  "is this value above or below a threshold?" over and over.
- The **Neural Network** boundary is smooth and curved -- a direct result of chaining weighted sums
  through non-linear activation functions.
- Neither shape is inherently correct. The right choice depends on whether the true market boundary is
  closer to a series of thresholds or a smooth transition.

</div>

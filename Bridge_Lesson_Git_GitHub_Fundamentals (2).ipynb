## 2. What Is Version Control?

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Your course folder already syncs to the cloud (Yandex Disk, Google Drive, OneDrive, ...). That protects you
from losing files, but it does **not** give you real version control:

| Cloud sync | Git |
|---|---|
| Keeps the latest version of each file | Keeps every meaningful snapshot you choose to save |
| "Undo" is limited and vague | You can jump back to the exact state after any commit |
| No record of *why* something changed | Every snapshot has a message explaining the change |
| Awkward or impossible to share just your code | `git push` puts your work on GitHub, shareable with one link |

**Git** is the tool that takes the snapshots, on your own computer. **GitHub** is a website that hosts a copy
of those snapshots online -- that hosted copy is what becomes your portfolio.

</div>

## 4. The Core Workflow: init, status, add, commit, log

We practice on a **scratch folder** first, so mistakes here cost nothing.


````python
import os

demo_dir = os.path.expanduser("~/git_demo")
os.makedirs(demo_dir, exist_ok=True)
os.chdir(demo_dir)
print(f"Working in: {os.getcwd()}")
!git init
````

`git init` turns this folder into a Git repository -- it creates a hidden `.git` folder that will hold
every snapshot from now on. Nothing is tracked yet.


````python
# Create a tiny file to track
with open("notes.md", "w") as f:
    f.write("# My ML4PEGs Notes\n\nStarting Git today.\n")

!git status
````

`git status` is your most-used command -- it always tells you what has changed and what to do next.
`notes.md` shows as **untracked**: Git sees it, but is not yet keeping snapshots of it.


````python
!git add notes.md
!git status
````

`git add` moves a file into the **staging area** -- the list of changes that will go into the *next*
snapshot. Staging is separate from committing on purpose: it lets you build a snapshot out of exactly the
files you want, even if you have other unfinished changes sitting in the folder.


````python
!git commit -m "Add first notes file"
!git log --oneline
````

`git commit` takes everything staged and saves it as a permanent, named snapshot. The `-m` message
should say **what** changed and briefly **why** -- future you (or a teammate) will read this, not just you today.

`git log --oneline` displays the commit history of your Git repository in a compact, one-line-per-commit format.


````python
# Make a change and repeat the loop
with open("notes.md", "a") as f:
    f.write("\nAdded a second line after learning git add/commit.\n")

!git status
!git add notes.md
!git commit -m "Expand notes with a second line"
!git log --oneline
````

## 5. The Three States

<div style="background: #f0f7ff; border-left: 5px solid #2d6a9f; padding: 16px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Every file in a Git repository is in one of three states, and the four commands above move it between them:

```
WORKING DIRECTORY  --git add-->  STAGING AREA  --git commit-->  REPOSITORY (.git history)
(your edits)                     (next snapshot,                (permanent, named
                                   in progress)                   snapshots)
```

`git status` always tells you which files are in which state. If you only remember one command, make it
that one -- run it constantly, it never changes anything, it only reports.

</div>

## 6. Writing a .gitignore for This Course

<div style="background: #fff8e1; border-left: 5px solid #e0a800; padding: 14px 20px; border-radius: 0 8px 8px 0; margin: 12px 0;">

Not everything belongs in Git. Large data files bloat the repository, and generated files
(`.ipynb_checkpoints`, `__pycache__`) just create noise. A `.gitignore` file tells Git to never track
matching files, even with `git add .`.

</div>


````python
gitignore_content = """# Data files (large -- keep these on cloud storage, not in Git)
data/
*.csv
*.las

# Jupyter checkpoint files
.ipynb_checkpoints/

# Python cache
__pycache__/
*.pyc
"""

with open(".gitignore", "w") as f:
    f.write(gitignore_content)

# Prove it works: create a fake data file and checkpoint folder, then check status
os.makedirs("data", exist_ok=True)
with open("data/well_log_data.csv", "w") as f:
    f.write("GR_API,NPHI_frac\n80,0.2\n")
os.makedirs(".ipynb_checkpoints", exist_ok=True)
with open(".ipynb_checkpoints/notes-checkpoint.ipynb", "w") as f:
    f.write("{}")

!git status
````

Notice `data/` and `.ipynb_checkpoints/` do **not** appear as untracked -- `.gitignore` is hiding them
from Git entirely. Only `.gitignore` itself shows up as a new file to add.


````python
!git add .gitignore
!git commit -m "Add .gitignore for data files and checkpoints"
!git log --oneline
````

## 7. Connecting to GitHub

Everything so far has been **local** -- snapshots live only on this computer. Connecting to GitHub gives you
a hosted copy, which is what turns your work into a shareable portfolio.

**Do this once, on github.com, outside this notebook:**

1. Create a free account at [github.com](https://github.com) if you do not have one.
2. Click **New repository**. Name it something like `ml4pegs-coursework`. Leave it empty (no README,
   no `.gitignore`) -- you already have local commits and do not want conflicting files.
3. Copy the repository URL GitHub shows you, e.g. `https://github.com/your-username/ml4pegs-coursework.git`.

**Then, back in a terminal (or a notebook cell), connect your local repo to it:**

```bash
git remote add origin https://github.com/your-username/ml4pegs-coursework.git
git push -u origin main
```

- `git remote add origin <url>` -- tells your local repo where its GitHub copy lives, and names it `origin`.
- `git push -u origin main` -- uploads your commits. `-u` remembers this pairing, so future pushes are just `git push`.
- GitHub will prompt you to sign in (a browser popup or a personal access token, depending on your setup).

We are not running these two commands in this notebook, since they need your real GitHub account --
run them in Section 8 with your actual repository.

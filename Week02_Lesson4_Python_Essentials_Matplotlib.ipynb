## 2. Matplotlib -- Plotting with Python

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

**Matplotlib** is Python's standard plotting library. The `pyplot` submodule gives you a simple, function-based interface.

```python
import matplotlib.pyplot as plt

plt.plot(x, y)          # draw a line
plt.xlabel("x label")
plt.ylabel("y label")
plt.title("My Plot")
plt.grid(True)
plt.show()              # display the figure
```

We will use NumPy arrays (from Lesson 3) as the data source.

</div>

````python
import numpy as np
import matplotlib.pyplot as plt

# Well_A GR log -- 21 depth samples, 0.5 m spacing
depth = np.arange(2000.0, 2010.5, 0.5)
gr    = np.array([128, 122, 115, 62, 45, 38, 42, 55, 80, 88,
                   60,  48,  35, 42, 118, 125, 130, 115, 95, 88, 82])

# Basic line plot
plt.plot(gr, depth)
plt.xlabel("GR (API)")
plt.ylabel("Depth (m)")
plt.title("Well_A -- Gamma Ray Log")
plt.grid(True)
plt.gca().invert_yaxis()   # depth increases downward
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

depth = np.arange(2000.0, 2010.5, 0.5)
gr    = np.array([128, 122, 115, 62, 45, 38, 42, 55, 80, 88,
                   60,  48,  35, 42, 118, 125, 130, 115, 95, 88, 82])
nphi  = np.array([0.08, 0.09, 0.10, 0.28, 0.32, 0.35, 0.30, 0.26,
                   0.15, 0.13, 0.25, 0.27, 0.31, 0.30, 0.10, 0.09,
                   0.08, 0.10, 0.12, 0.13, 0.14])

# Two panels sharing the depth axis
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(7, 6), sharey=True)

# GR track
ax1.plot(gr, depth, color="#2563eb", linewidth=1.8)
ax1.axvline(x=75, color="gray", linestyle="--", linewidth=0.8, label="cutoff")
ax1.fill_betweenx(depth, gr, 75, where=(gr < 75), color="#f0c040", alpha=0.3)
ax1.set_xlabel("GR (API)"); ax1.set_title("GR"); ax1.set_xlim(0, 160)
ax1.legend(fontsize=8)

# NPHI track
ax2.plot(nphi, depth, color="#2ca87f", linewidth=1.8)
ax2.set_xlabel("NPHI (V/V)"); ax2.set_title("NPHI"); ax2.set_xlim(0, 0.5)

for ax in (ax1, ax2):
    ax.invert_yaxis()
    ax.grid(True, linestyle=":", alpha=0.5)

ax1.set_ylabel("Depth (m)")
fig.suptitle("Well_A -- GR and NPHI Log Display", fontweight="bold")
plt.tight_layout()
plt.show()
````

> **Key Matplotlib patterns used throughout the course:**
> - `fig, axes = plt.subplots(rows, cols, ...)` -- create multi-panel figure
> - `ax.invert_yaxis()` -- depth increases downward (petrophysics convention)
> - `ax.fill_betweenx(y, x1, x2, where=...)` -- shade sand intervals
> - `sharey=True` -- panels share the depth axis

## 3. Python Dictionaries

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A **dictionary** stores data as `key: value` pairs inside curly braces `{}`.
You look up data by name (key) rather than by position (index).

```python
d = {"key1": value1, "key2": value2}
d["key1"]        # access a value
d["new"] = val   # add a new pair
"key1" in d      # True / False
```

</div>

````python
# A well summary stored as a dictionary
well = {
    "name":       "Well_A01",
    "depth_m":    2500,
    "field":      "North Shelf Block 7",
    "drilled":    True,
}

print(well)
print("Well name :", well["name"])
print("Depth     :", well["depth_m"], "m")

# Add a new entry
well["operator"] = "PetroCo"
print("Updated   :", well)

# Check if a key exists
print("'field' in well?   ", "field"   in well)
print("'GR_API' in well?  ", "GR_API"  in well)
````

````python
# When each value is a LIST, the dictionary looks exactly like a table.
# This is the structure Pandas uses to build a DataFrame -- you will see it in Lesson 5.

log_table = {
    "Depth_m":    [2000.0, 2000.5, 2001.0, 2001.5, 2002.0],
    "GR_API":     [128,    122,    115,    62,     45    ],
    "NPHI_frac":  [0.08,   0.09,   0.10,   0.28,   0.32  ],
}

print("Keys  :", list(log_table.keys()))
print("Depths:", log_table["Depth_m"])
print("GR    :", log_table["GR_API"])
print("Rows  :", len(log_table["Depth_m"]))
````

> **Remember this structure** -- in Lesson 5 you will pass a dictionary exactly like `log_table` to `pd.DataFrame()` and get a spreadsheet-like table back.

## 4. if / elif / else

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

`if / elif / else` lets your code make **decisions** based on a condition.

```python
if condition:
    # runs when condition is True
elif another_condition:
    # runs when first is False but this is True
else:
    # runs when all conditions are False
```

Comparison operators: `<` `>` `<=` `>=` `==` `!=`
Combine with: `and`, `or`

</div>

````python
gr_value = 62   # change this and re-run

if gr_value < 75:
    print("Sand -- below GR cutoff")
else:
    print("Shale -- above GR cutoff")

# Combining two conditions with 'and'
nphi_value = 0.29
if gr_value < 75 and nphi_value > 0.25:
    print("Good reservoir: sand + high porosity")
else:
    print("Not a good reservoir")
````

````python
# Multiple branches with elif -- GR quality classifier
gr_value = 62

if gr_value < 50:
    quality = "Good sand"
elif gr_value < 75:
    quality = "Marginal sand"
else:
    quality = "Shale"

print(f"GR = {gr_value} API  -->  {quality}")

# Test a few more values
for_check = [35, 62, 90, 42, 130]
for gr in for_check:
    if gr < 50:
        q = "Good sand"
    elif gr < 75:
        q = "Marginal sand"
    else:
        q = "Shale"
    print(f"  GR={gr:4d}  -->  {q}")
````

> **The GR cutoff (75 API)** is a standard industry threshold. Values below 75 API indicate sand-dominated intervals; above indicates shale. Your specific field may use a different cutoff.

## 5. for Loops

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A `for` loop repeats a block of code **once for each item in a list**.

```python
for item in my_list:
    # this runs once per item
```

The loop variable (`item`) takes the value of each element in turn.

</div>

````python
# Loop over a list of GR readings from one well
gr_readings = [128, 62, 45, 38, 55, 80, 42, 118]

print("GR readings for Well_A:")
for gr in gr_readings:
    print(f"  GR = {gr} API")

# Count how many readings are below the sand cutoff
sand_count = 0
for gr in gr_readings:
    if gr < 75:
        sand_count = sand_count + 1

print(f"\nSand intervals (GR < 75): {sand_count} out of {len(gr_readings)}")
````

````python
# Build a classification list by looping and using if/elif
gr_readings = [128, 62, 45, 38, 55, 80, 42, 118]
depths      = [2000.0, 2001.0, 2002.0, 2002.5, 2003.0, 2004.0, 2005.0, 2006.0]
lithologies = []    # empty list -- filled inside the loop

for gr in gr_readings:
    if gr < 50:
        lithologies.append("Good sand")
    elif gr < 75:
        lithologies.append("Marginal sand")
    else:
        lithologies.append("Shale")

print("Lithology list:", lithologies)
print()

# Count each category
sand_count = lithologies.count("Good sand") + lithologies.count("Marginal sand")
print(f"Sand intervals  : {sand_count}")
print(f"Shale intervals : {lithologies.count('Shale')}")
````

> **Loop + if is a fundamental pattern** for processing log data row by row.
> In Lesson 5, Pandas `.apply()` does the same thing -- but faster and in one line, using a function or lambda you write once.

## 6. Python Functions

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A **function** packages a block of code so you can run it again and again with different inputs.

```python
def function_name(parameter):
    # do something with parameter
    return result
```

- `def` starts the definition
- `parameter` is the input (one in this lesson)
- `return` sends the result back
- Call it with `function_name(value)`

</div>

````python
# A function that converts depth from metres to feet
def depth_to_feet(depth_m):
    return depth_m * 3.28084

# Call the function with different inputs
print(depth_to_feet(100))     # 328.084
print(depth_to_feet(2000))    # 6561.68
print(depth_to_feet(2500))    # 8202.1

# Use it in a loop
depths_m = [2000.0, 2001.0, 2002.0, 2003.0]
for d in depths_m:
    print(f"{d} m  -->  {depth_to_feet(d):.2f} ft")
````

````python
# A function that classifies a GR value -- uses if/elif inside
def classify_gr(gr):
    if gr < 50:
        return "Good sand"
    elif gr < 75:
        return "Marginal sand"
    else:
        return "Shale"

# Test on individual values
print(classify_gr(35))    # Good sand
print(classify_gr(62))    # Marginal sand
print(classify_gr(130))   # Shale

# Use it in the loop from Section 5 -- now much cleaner
gr_readings = [128, 62, 45, 38, 55, 80, 42, 118]
print()
for gr in gr_readings:
    print(f"  GR={gr:4d}  -->  {classify_gr(gr)}")
````

> **Compare Sections 5 and 6:** the loop in Section 5 repeated the if/elif logic inside every loop.
> Here, `classify_gr()` is defined once and called everywhere -- if the cutoff ever changes, you update one line.

## 7. Lambda Functions

<div style="background:#f0f7ff;border-left:5px solid #2d6a9f;padding:14px 20px;border-radius:0 8px 8px 0;margin:10px 0;">

A **lambda** is a one-line shorthand for a simple function. Same result, less typing.

```python
# Regular function (3 lines)
def square(x):
    return x ** 2

# Lambda (1 line) -- no def, no return
square = lambda x: x ** 2
```

Syntax: `lambda parameter: expression`

</div>

````python
# Regular function
def to_feet(m):
    return m * 3.28084

# Same thing as a lambda
to_feet_lam = lambda m: m * 3.28084

print(to_feet(100))        # 328.084
print(to_feet_lam(100))    # 328.084  -- identical result

# Lambda with a conditional expression
# Syntax: lambda x: value_if_true  if  condition  else  value_if_false
classify = lambda gr: "Sand" if gr < 75 else "Shale"

print(classify(45))    # Sand
print(classify(120))   # Shale
````

````python
# You can use a lambda directly inside a loop (no named function needed)
gr_readings = [128, 62, 45, 38, 55, 80, 42, 118]

classify_lam = lambda gr: "Good sand" if gr < 50 else ("Marginal sand" if gr < 75 else "Shale")

print("Lambda classification:")
for gr in gr_readings:
    print(f"  GR={gr:4d}  -->  {classify_lam(gr)}")

# In Lesson 5, this exact pattern is used inside Pandas .apply():
#   df["Lithology"] = df["GR_API"].apply(lambda gr: "Sand" if gr < 75 else "Shale")
````

> **Rule of thumb:**
> - Use a **named function** (`def`) when the logic has multiple lines or you reuse it in many places.
> - Use a **lambda** for short, one-line transformations used once -- especially inside `.apply()` in Lesson 5.
